Project Description:
This project implements an end-to-end data analysis pipeline to analyze a real-world used car dataset.
Using Pandas for data manipulation, NumPy for mathematical operations, and Matplotlib for data visualization,
the system loads the data, performs data exploration, handles missing values, computes key statistical metrics,
and generates charts to uncover underlying market patterns and trends.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

print("--- 1. Loading the Dataset ---")
# Reading the CSV dataset
df = pd.read_csv("cars_data.csv")

print("\n* First 10 rows of the dataset:")
display(df.head(10))

print("\n* Last 10 rows of the dataset:")
display(df.tail(10))
```

 Explanation & Insights (Step 1: Load the Data):
* Code Explanation: In this cell, we import the primary data science libraries: pandas, numpy, and matplotlib.pyplot. We use pd.read_csv() to load the dataset into a DataFrame named df. Then, we use df.head(10) and df.tail(10) to display the first and last 10 rows of the dataset to understand its initial layout.
* Observed Patterns: The initial inspection shows that the dataset contains detailed records of used cars, including architectural attributes like vehicle names, manufacturing years, selling prices, present prices, kilometers driven, fuel types, transmission mechanisms, and owner history.

```python
print("--- 2. Data Exploration ---")

# Number of rows and columns
rows, cols = df.shape
print(f"* Number of Rows: {rows} | Number of Columns: {cols}")

# Data types of each column
print("\n* Data types:")
print(df.dtypes)

# Check for missing values
print("\n* Checking for missing values:")
print(df.isnull().sum())
```

Explanation & Insights (Step 2: Data Exploration):
* Code Explanation: We use df.shape to extract the dimensions of the dataset (rows and columns). df.dtypes is used to check the data types of each attribute (e.g., integers, floats, or objects). Finally, df.isnull().sum() is executed to calculate the exact number of missing values per column to evaluate data completeness.
* Exploration Insights: The structural output confirms the dataset's size and shows a combination of continuous numerical features (prices, kilometers) and categorical features (fuel type, transmission). The data scan reveals that there are zero missing values across all columns, indicating a clean data structure.

```python
print("--- 3. Data Cleaning ---")

# Check and handle missing values dynamically
missing_count = df.isnull().sum().sum()
if missing_count > 0:
    print(f"Found {missing_count} missing values. Cleaning dataset...")
    for col in df.columns:
        if df[col].isnull().sum() > 0:
            if df[col].dtype == "object":
                df[col].fillna(df[col].mode()[0], inplace=True)
            else:
                df[col].fillna(df[col].mean(), inplace=True)
else:
    print("Data cleaning checked: No missing values detected. Dataset is fully optimized.")
```

 Explanation & Insights (Step 3: Data Cleaning):
* Code Explanation: To strictly fulfill the data cleaning requirements, we implemented an automated verification block. It dynamically checks if any missing entries exist. If missing values are found, it fills categorical missing values with the mode and numerical missing values with the mean.
* Cleaning Insights: Since our dynamic check returned 0 missing values, no structural modifications were forced upon the dataset. The pipeline confirms that the original car rows are highly reliable, structurally intact, and ready for accurate mathematical analysis.

```python
print("--- 4. Statistical Analysis ---")

# Selecting numerical columns for calculations
numeric_cols = ["Selling_Price", "Present_Price", "Kms_Driven"]

for col in numeric_cols:
    print(f"\n[ Statistics for: {col} ]")
    print(f"  - Mean: {df[col].mean():.2f}")
    print(f"  - Median: {df[col].median():.2f}")
    print(f"  - Standard Deviation: {df[col].std():.2f}")
    print(f"  - Minimum Value: {df[col].min():.2f}")
    print(f"  - Maximum Value: {df[col].max():.2f}")
```

 Explanation & Insights (Step 4: Statistical Analysis):
* Code Explanation: We utilize built-in Pandas Series mathematical operations (`.mean()`, .median(), .std(), .min(), .max()`) on the core numerical features: `Selling_Price, Present_Price, and Kms_Driven. This method ensures computational precision and avoids missing structural outputs.
* Statistical Insights & Relationships: * Prices: The statistical distribution shows that the average Present_Price (original showroom price) is noticeably higher than the average Selling_Price (used market price), reflecting clear asset depreciation over time.
  * Variance: The relatively high standard deviation in Kms_Driven indicates a diverse market selection, ranging from nearly new vehicles to heavily driven cars.

```python
print("--- 5. Data Visualization: Plot 1 ---")

# Plot 1: Bar Chart (Average Selling Price by Fuel Type)
plt.figure(figsize=(8, 5))
fuel_prices = df.groupby("Fuel_Type")["Selling_Price"].mean()
plt.bar(fuel_prices.index, fuel_prices.values, color=["orange", "blue", "green"][:len(fuel_prices)])
plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

 Explanation & Insights (Plot 1: Bar Chart):
* Code Explanation: This block aggregates the dataset using groupby('Fuel_Type') and calculates the average Selling_Price for each category. We then build a bar chart using plt.bar() to contrast these values visually.
* Visual Insights & Patterns: The bar chart reveals a substantial pricing pattern based on fuel categories. Cars running on Diesel exhibit a significantly higher average selling price compared to Petrol-fueled vehicles. This pattern suggests that diesel cars retain market value better, likely due to fuel efficiency or heavy-duty engine profiles.

```python
print("--- 5. Data Visualization: Plot 2 ---")

# Plot 2: Line Chart (Trend of Average Selling Price over Years)
plt.figure(figsize=(8, 5))
yearly_prices = df.groupby("Year")["Selling_Price"].mean().sort_index()
plt.plot(yearly_prices.index, yearly_prices.values, marker="o", color="green", linestyle="-", linewidth=2)
plt.title("Trend of Average Selling Price over Years")
plt.xlabel("Year")
plt.ylabel("Average Selling Price")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

 Explanation & Insights (Plot 2: Line Chart):
* Code Explanation: We group the car data by manufacturing Year and compute the mean selling price over time. plt.plot() generates a chronological trendline decorated with circular markers to depict price velocity across historical models.
* Visual Insights & Trends: The line chart reveals a clear, upward positive correlation between the car's manufacturing year and its resale market value. Newer models (higher manufacturing years) consistently command higher prices, illustrating how vehicle value drops with age and older models depreciate.

```python
print("--- 5. Data Visualization: Plot 3 ---")

# Plot 3: Histogram (Distribution of Kilometers Driven)
plt.figure(figsize=(8, 5))
plt.hist(df["Kms_Driven"], bins=15, color="purple", edgecolor="black", alpha=0.7)
plt.title("Distribution of Kilometers Driven (Kms_Driven)")
plt.xlabel("Kilometers Driven")
plt.ylabel("Number of Cars")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

 Explanation & Insights (Plot 3: Histogram):
* Code Explanation: We utilize plt.hist() split into 15 specific intervals (bins) to track the density distribution of the Kms_Driven column across the entire collection.
* Visual Insights & Key Conclusions: The histogram shows a right-skewed distribution. The vast majority of used cars listed in the marketplace fall within the lower range of total mileage (concentrated below 50,000 kilometers). This confirms that the market heavily features relatively low-mileage vehicles, which strongly influences the overall pricing environment.
